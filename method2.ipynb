```python
import pandas as pd
import numpy as np

from mlxtend.frequent_patterns import apriori, association_rules

import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', 50)
sns.set(style='whitegrid')
```

```python
df = pd.read_csv('datasets/Final_Augmented_dataset_Diseases_and_Symptoms.csv')

df.head()
```

```python
print("Shape:", df.shape)
print("Columns", df.columns[:5])
```

```python
df.fillna(0, inplace=True)

binary_cols = df.columns[1:]
df[binary_cols] = df[binary_cols].applymap(lambda x: 1 if x == 1 else 0)

symptom_df = df.drop('diseases', axis=1)

print("Symptom matrix shape:", symptom_df.shape)
symptom_df.head()
```

```python
frequent_itemsets = apriori(symptom_df, min_support=0.05, use_colnames=True)

positive_rules = association_rules(frequent_itemsets, metric="lift", min_threshold=1.0)

positive_rules = positive_rules[positive_rules['lift'] > 1].sort_values(by='lift', ascending=False)

print("Top 10 Positive Association Rules:")
positive_rules.head(10)
```

```python
# don't run the above code cell ( generated by gpt) - for negative association rules first reduce the batch size then work on the code bellow after that complete rest of the eda works 


from mlxtend.frequent_patterns import fpgrowth

# Efficient inversion and type conversion
neg_df = ~symptom_df.astype(bool)

# Keep only common symptoms
symptom_freq = symptom_df.mean()
common_symptoms = symptom_freq[symptom_freq >= 0.05].index
neg_df_reduced = neg_df[common_symptoms]

# Run FP-Growth (more efficient than Apriori)
neg_itemsets = fpgrowth(neg_df_reduced, min_support=0.2, use_colnames=True)

# Generate rules
neg_rules = association_rules(neg_itemsets, metric="lift", min_threshold=1.0)

# Filter and display
negative_rules = neg_rules[neg_rules['lift'] > 1].sort_values(by='lift', ascending=False)
print("Top 10 Negative Association Rules:")
negative_rules.head(10)
```
