```python
import numpy as np
import pandas as pd

from keras import models
from keras import layers

from sklearn.preprocessing import OneHotEncoder
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv('datasets/Final_Augmented_dataset_Diseases_and_Symptoms.csv')

df.head()
```

```python
target = df.pop("diseases")

len(target.unique())

enc = OneHotEncoder(handle_unknown='ignore')

test = target.to_numpy().reshape(-1, 1)

encoded_lables = enc.fit_transform(test)

encoded_lables_arr = encoded_lables.toarray()
```

```python
X = df.to_numpy()
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, encoded_lables_arr, test_size=0.33, random_state=42)
```

```python
model = models.Sequential([
    layers.Dense(1024, activation="relu", input_shape=(X_train.shape[1],)),
    layers.Dense(1024, activation="relu"),
    layers.Dense(773, activation="softmax")
])
```

```python
model.compile(optimizer="rmsprop", loss="categorical_crossentropy", metrics=['accuracy'])
```

```python
model.fit(X_train, y_train, batch_size=1000, epochs=5)
```

```python
accuracy, loss = model.evaluate(X_test, y_test)
```
